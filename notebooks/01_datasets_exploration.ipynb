# 1. Exploratory data analysis

## Dataset collection

```python
from src.paths import data_dir, somermeyer_data_dir, tsuboyama_data_dir
import shutil
import pandas as pd

src_folder = data_dir / "zero_shot_substitutions_scores"

# Somermeyer
for file in src_folder.glob("*Somermeyer*"):
    shutil.move(file, somermeyer_data_dir / file.name)

# Tsuboyama
for file in src_folder.glob("*Tsuboyama*"):
    df = pd.read_csv(tsuboyama_data_dir / file)
    df["num_mutations"] = df["mutant"].apply(lambda x: len(x.split(":")))
    if df.num_mutations.nunique() > 1:
        shutil.move(file, tsuboyama_data_dir / file.name)
    else:
        print(f"File {file.name} has only single mutants. Skipping")
```

## Distribution of mutations

```python
import matplotlib.pyplot as plt
from src.paths import somermeyer_data_dir, main_figures_dir
from src.plotting import mut_hist_somermeyer

src_dir = somermeyer_data_dir
files = sorted(src_dir.glob("*.csv"))

fig, axs = plt.subplots(nrows=3, ncols=1, figsize=(6, 12))

for ax, file in zip(axs, files):
    mut_hist_somermeyer(file, ax=ax)

plt.tight_layout()
out_file = main_figures_dir / "1a_somermeyer_mutations_hist.png"
fig.savefig(out_file, dpi=300, bbox_inches="tight")
plt.close(fig)
```

## Distribution of Tsuboyama values

```python
from src.utils import convert_name_to_gfp
from src.paths import tsuboyama_data_dir
from pathlib import Path
import pandas as pd

num_genotypes = []
for file in Path(tsuboyama_data_dir).glob("*.csv"):
    df = pd.read_csv(file)
    name = "_".join(file.stem.split("_")[:2])
    num_genotypes.append(df.shape[0])
print("Minimal number of genotypes in Tsuboyama datasets: {}".format(min(num_genotypes)))
print("Maximal number of genotypes in Tsuboyama datasets: {}".format(max(num_genotypes)))
print("Total number of genotypes in Tsuboyama datasets: {}".format(sum(num_genotypes)))
```
