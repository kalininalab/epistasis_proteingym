## Tsuboyama datasets

```python
import os
from pathlib import Path
import pandas as pd
from src.paths import data_dir, tsuboyama_data_dir, interm_tables_dir
from src.data_processing import add_dg

# dG values
dg_values = pd.read_csv(os.path.join(data_dir, "Tsuboyama2023_Dataset2_Dataset3_20230416.csv"), low_memory=False)
# Iterate over all CSV files in the Tsuboyama folder
for file in os.listdir(tsuboyama_data_dir):
    # Read each CSV file into a DataFrame
    file_path = os.path.join(tsuboyama_data_dir, file)
    dataframe = pd.read_csv(file_path)
    name = file.split(".")[0].split("_")[-1]
    # Process the DataFrame to add deltaG values
    df = add_dg(dataframe, name, dg_values)
    # Save the processed DataFrame to a new CSV file
    dg_folder = Path(interm_tables_dir, "tsuboyama")
    dg_folder.mkdir(parents=True, exist_ok=True)
    df.to_csv(dg_folder / file)
```

```python
import subprocess, shlex
from pathlib import Path
from src.paths import interm_tables_dir, external_dir

def run_in_conda_env(env_name: str, script: Path, **kwargs):
    args = []
    for k, v in kwargs.items():
        args.extend([f"--{k.replace('_','-')}", str(v)])
    cmd = ["conda", "run", "-n", env_name, "python", str(script), *args]
    print("[run]", " ".join(shlex.quote(c) for c in cmd))
    subprocess.run(cmd, check=True)

# define input and output folders
in_dir = interm_tables_dir / "tsuboyama"
out_dir   = interm_tables_dir / "tsuboyama_recon_dg"
out_dir.mkdir(parents=True, exist_ok=True)

# now call the external script in the tsuboyama env
run_in_conda_env(
    "protease-pipeline",                                    # the conda environment name
    external_dir / "tsuboyama/additive_model.py",   # path to the script
    **{"in": in_dir, "out": out_dir}            # CLI args for the script
)
```
